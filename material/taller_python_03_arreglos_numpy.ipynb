# Arreglos

Un **arreglo** (o array) es una estructura de datos que almacena elementos del mismo tipo en una secuencia ordenada. En Python, el manejo de arreglos se logra mediante el uso del módulo `numpy`. A diferencia de las listas de Python, los arreglos de **NumPy**:

- **Son homogéneos**: Todos los elementos son del mismo tipo de dato
- **Son eficientes**: Operaciones vectorizadas más rápidas
- **Ocupan menos memoria**: Almacenamiento contiguo en memoria
- **Soportan operaciones matemáticas**: Element-wise operations

Las listas de Python son flexibles pero lentas para operaciones matemáticas. NumPy proporciona:
- **Performance**: 10-100x más rápido que listas de Python
- **Operaciones vectorizadas**: Sin necesidad de bucles explícitos
- **Interfaz estándar**: Usado por pandas, matplotlib, scikit-learn, etc.

```python
# Importar NumPy
import numpy as np
```


### Dimensiones de un arreglo
- **1D**: Vector (fila de números)
- **2D**: Matriz (filas y columnas)
- **3D+**: Tensor (múltiples dimensiones)

### Atributos Importantes
- `dtype`: Tipo de dato de los elementos
- `size`: Número total de elementos
- `shape`: Dimensiones del arreglo

## Arreglos Unidimensionales

### Crear arreglos vacíos

```python
a = np.zeros(10, dtype='int')
print(a)
```

```python
b = np.ones(5, dtype='float')
print(b)
```

### Crear arreglos con valores equiespaciados

`np.arange(start, stop, step)` -> desde `start` a `stop` sin incluirlo

```python
c = np.arange(3.0, 9.0, 0.5)  # Funciona igual que el range pero los intervalos pueden ser reales.
print(c)
```

`linspace(inicio, termino, cantidad de elementos)`

```python
d = np.linspace(1, 8, 5)  # Crea arreglo de valores equiespaciados. (Stop - Start) / n-1
print(d)
```

## Arreglos Bidimensionales

```python
print(a)
```

```python
a.shape
```

```python
len(a)
```

```python
c = np.zeros((3, 4))
print(c)
```

```python
d = np.random.random((3, 3))
print(d)
```

```python
d[1, 2]
```

```python
for i in range(d.shape[0]):
    for j in range(d.shape[1]):
        print(d[i, j])
```

## Operaciones Matriciales

```python
a = np.arange(8)
print(a)
```

```python
b = a.reshape((2, 4))
print(b)
```

```python
b.T
```

```python
np.transpose(b)
```

```python
# Reducciones
print(b.min(axis=0))
```

```python
print(b.sum(axis=1))
```

```python
b.flatten()
```

```python
# Dot
m = np.array([[2, 1], [4, 5]])
n = np.array([[5], [8]])

print(f"m: {m}")
print(f"n: {n}")

np.dot(m, n)
```
